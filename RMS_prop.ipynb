```python
#  RMSprop
# ============================================

import numpy as np
import matplotlib.pyplot as plt
```

```python


# --------------------------------------------
# 1. Dataset
# --------------------------------------------

np.random.seed(42)

X = np.linspace(-5, 5, 100)
y = 3 * X + 2 + np.random.randn(100) * 0.5

# --------------------------------------------
# 2. Parameters
# --------------------------------------------

w = 0.0
b = 0.0

learning_rate = 0.01
beta = 0.9
epsilon = 1e-8
epochs = 100

# Moving averages
s_w = 0.0
s_b = 0.0

loss_history = []

# --------------------------------------------
# 3. RMSprop Training
# --------------------------------------------

for epoch in range(epochs):

    # Prediction
    y_pred = w * X + b

    # Error
    error = y_pred - y

    # Loss
    loss = np.mean(error ** 2)
    loss_history.append(loss)

    # Gradients
    dw = np.mean(2 * X * error)
    db = np.mean(2 * error)

    # Moving average of squared gradients
    s_w = beta * s_w + (1 - beta) * (dw ** 2)
    s_b = beta * s_b + (1 - beta) * (db ** 2)

    # RMSprop update
    w = w - learning_rate * dw / (np.sqrt(s_w) + epsilon)
    b = b - learning_rate * db / (np.sqrt(s_b) + epsilon)


# --------------------------------------------
# 4. Results
# --------------------------------------------

print("RMSprop Results")
print("Weight:", w)
print("Bias:", b)
print("Final Loss:", loss_history[-1])


# --------------------------------------------
```

```python
# 5. Plot
# --------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("RMSprop Training")
plt.grid()

plt.show()
```
