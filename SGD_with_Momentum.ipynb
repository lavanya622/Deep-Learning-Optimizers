```python
# SGD WITH MOMENTUM
# ============================================

import numpy as np
import matplotlib.pyplot as plt
```

```python

# --------------------------------------------
# 1. Create simple dataset
# --------------------------------------------

np.random.seed(42)

X = np.linspace(-5, 5, 100)
y = 3 * X + 2 + np.random.randn(100) * 0.5

# --------------------------------------------
# 2. Initial parameters
# --------------------------------------------

w_sgd = 0.0
b_sgd = 0.0

w_momentum = 0.0
b_momentum = 0.0

# --------------------------------------------
# 3. Hyperparameters
# --------------------------------------------

learning_rate = 0.01
momentum = 0.9
epochs = 100

# Velocity
velocity_w = 0.0
velocity_b = 0.0

# Store losses
loss_sgd = []
loss_momentum = []

# --------------------------------------------
# 4. SGD
# --------------------------------------------

for epoch in range(epochs):

    # Prediction
    y_pred = w_sgd * X + b_sgd

    # Error
    error = y_pred - y

    # Loss
    loss = np.mean(error ** 2)
    loss_sgd.append(loss)

    # Gradients
    dw = np.mean(2 * X * error)
    db = np.mean(2 * error)

    # Parameter update
    w_sgd = w_sgd - learning_rate * dw
    b_sgd = b_sgd - learning_rate * db


# --------------------------------------------
# 5. SGD with Momentum
# --------------------------------------------

for epoch in range(epochs):

    # Prediction
    y_pred = w_momentum * X + b_momentum

    # Error
    error = y_pred - y

    # Loss
    loss = np.mean(error ** 2)
    loss_momentum.append(loss)

    # Gradients
    dw = np.mean(2 * X * error)
    db = np.mean(2 * error)

    # Momentum update
    velocity_w = momentum * velocity_w - learning_rate * dw
    velocity_b = momentum * velocity_b - learning_rate * db

    # Update parameters
    w_momentum = w_momentum + velocity_w
    b_momentum = b_momentum + velocity_b


# --------------------------------------------
# 6. Print results
# --------------------------------------------

print("SGD")
print("Weight:", w_sgd)
print("Bias:", b_sgd)
print("Final Loss:", loss_sgd[-1])

print("\nSGD with Momentum")
print("Weight:", w_momentum)
print("Bias:", b_momentum)
print("Final Loss:", loss_momentum[-1])


```

```python
# --------------------------------------------
# 7. Compare Loss
# --------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(loss_sgd, label="SGD")
plt.plot(loss_momentum, label="SGD + Momentum")

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("SGD vs SGD with Momentum")
plt.legend()
plt.grid()

plt.show()
```
